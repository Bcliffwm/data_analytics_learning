# What is a String?
## A string is a sequence of text made up of individual characters. Strings can be *indexed* to find certain characters or specific patterns

```python
text = "Hello World!"

print(text[4]) # printing the fifth character in the string sequence
print(text[0:2]) # printing the first three characters in the string sequence
print(text[::-1]) # reversing the string
```

```python
print(dir(text)) # printing the methods associated with the string class
```

```python
# Examples using string methods
# using the list provided by dir method to identify useful methods in string class
print([method.startswith('__') for method in dir(text)])

# using a conditional method of string class to evaluate each character
print([word.isspace() for word in text])
```

## Practical data manipulation example using strings

```python
running_time = '1:30:45'
print("Calling split method: {}".format(running_time.split(':'))) # using format method from string class

def convert_str_time(time_str: list[str]): # taking function from the coding problem challenge
    # using split method on input param - iterating over each string component in list comprehension
    splits = [time.split(':') for time in time_str]
    converted_seconds = [] # initializing empty list for handling time conversions to seconds
    for time in splits: # creating the loop for manipulating each string in the splits list
        seconds = 0 # rolling calculation for the converted times
        # using the map func to convert each hr/min/sec into integers and store in list
        hr_min_sec = list(map(int, time)) 
        # Reversing the list using slicing to compute length and iterate for time conversions correctly
        reversed_time = hr_min_sec[::-1]
        
        for i in range(len(reversed_time)):
            # applying chain rule to each integer in the reversed list to calculate the total time in seconds
            seconds += reversed_time[i] * (60 ** i)
            # showing how each integer element in reversed_time gets processed with f-string
            print(f'Processing: {reversed_time[i]}, total seconds: {seconds}') 
        converted_seconds.append(seconds) # appending the converted time to the empty list
    return converted_seconds

print(convert_str_time([running_time])) # calling the function
print((1 * 60**2) + (30 * 60**1) + (45 * 60**0)) # checking the output result
```

# String formatting
## String formatting is the process of putting your variables stored in memory into a string object
#### There are three options for formatting strings:
1. % operator
2. format method
3. f-strings

```python
# using the % operator from the Python Bible pg 60
# To use the % operator, you need to add the correct placeholder for your formatted string

name, age = "Ben", 30
print("My name is %s" % name) # adding the string placeholder in the string object
print("I am %d years old!" % age) # adding the integer placeholder in the string object
```

```python
# using the format method using same variables
print("My name is {} and I am {} years old!".format(name, age)) #implicit ordering of formatting

print("I am {1} years old and oh yeah, my name is {0}".format(name, age)) #explicit ordering of formatting

print("My legal name is {ln}, but people call me {n}. I'm {a} years old.".format(ln='Benjamin',n=name, a=age)) #keyword arguments

data = (name, age) # creating a tuple using the name and age variables above
print("Name: {}, Age: {}".format(*data)) # unpacking the data tuple with the format method
```

```python
# Using f-strings
print(f"Bonjour, je m'appelle {name}. J\'ai {age} ans.") # applying keywords with curly braces
```

# Finding text
## The find method supports regular expressions, but you can also import the re library for more flexible matches

```python
print(text.find('W')) # finding position of a specific character
print(text.find('Hello')) # finding position of a specific word
print(text.find('Why')) # if searching for substring that doesn't exist, -1 is returned
```

## Using regular expressions
### When using the re module in Python, use raw strings -> r''

```python
import re
print(re.match(r'[A-Z]', text).pos) # returning position of the first instance of a capital letter in string
print(re.findall(r'[A-Z]', text)) # returning all uppercase letters in text
print(re.findall(r'[a-z]', text)) # returning all lowercase letters in text
print(re.findall(r'[\D]', "Hello, World 123456")) # returning all characters that are non-digits
print(re.findall(r'\W', text+"123")) # returning all characters that are not alphanumeric
```

```python
# updating text with re library - see example on page 149 of Python Library
ages_text = """Ben is 30 years old, Jason is 29 years old, and Hannibal is 28 years old. Old, hoes!"""
print(re.sub(r'\d{1,3}', "100", ages_text)) # replacing each digit matched string with 100
```

### ETL/data processing example

```python
import string

# Raw, messy user input
raw_input = "   Nolan, Christopher!   "

# Cleaned data: "nolan christopher"
clean_input = raw_input.strip().lower().translate(str.maketrans("", "", string.punctuation))

print("Raw input:", raw_input, sep=" ")
print("Cleaned input:", clean_input, sep=" ")
```

### Log Parsing and File Ingestion example

```python
log_line = "2026-10-12 16:20:05 ERROR Database connection failed"

# Extracting the timestamp and log level
if log_line.startswith("2026"):
    parts = log_line.split(" ")
    log_level = parts[2]  # "ERROR"

print(parts)
print(log_level)
```
