# What is a List?
## A list is an *ordered* sequence of elements that are accessible and iterable via the *index*. The index is another list that holds a positional value associated with each element of the list for retrieval and sorting purposes.

#### Lists can hold multiple primative data types and composite data types
#### List indexes are 0 indexed - meaning that you always start with 0

# Accessing Lists - Slicing and Indexing

```python
my_list = ['1', 2, 'a', 5.6, 'New York']

print(my_list) # printing the entire list
print(my_list[0]) # printing the first element of the list
print(my_list[1:3]) # printing a subset of the list
print(my_list[-1]) # printing the last element of the list
```

```python
[type(elem) for elem in my_list] # printing each element's data type using list comprehension
```

## List Slicing

* first index value is starting position in a list (inclusive)
* second index value is last position in a list (not inclusive)
* third index value is the skip value

```python
print(my_list[0:4]) # first position of list, fourth position of list
print(my_list[1:5]) # second position of list, fifth position of list
print(my_list[:3]) # grabbing everything before the 3rd element of the list (inclusive)
print(my_list[1:]) # grabbing everything after the 2nd element of the list (inclusive)
print(my_list[0:5:2]) # first position of list, fifth position of list, skip by 2
print(my_list[5:0:-1]) # fifth position of list, first position of list, skip by -1
```

```python
print(my_list.index(2)) # prints position of the value provided - in this case 1
print(my_list.index('New York'))
try:
    print(my_list.index('Virginia')) # if value provided is not in list, then ValueError is raised
except ValueError:
    print('Value is not in my_list')
```

# Useful List Methods

```python
# printing the list methods
print([prop for prop in dir(my_list) if prop[0] != '_'])
```

## Adding Values to a List

### Append Method

```python
my_list.append(['Hannibal', 'Jason', 'Ross']) # append adds element to a list at the very end - notice how lists can also be added to lists
print(my_list)
```

```python
# initializing empty list to store an array of values
numbers_list = []
curr_num = 0 # initializing int var to add values to list

# Using a while loop to append values to numbers_list
while curr_num <= 50:
    numbers_list.append(curr_num) # adding current number in the loop to list
    curr_num += 5 # adding 5 to the current number
```

### Insert Method

```python
# takes a position value and object as arguments then inserts object at specific position, updating list values and index
my_list.insert(3, 'Trade')
print(my_list)
```

## Mathematical Operations for lists

```python
print(numbers_list)
print('Length operation:', len(numbers_list)) # returns the count of items in a list
print('Minimum operation:', min(numbers_list)) # returns minimum value from list
print('Maximum operation:', max(numbers_list)) # returns maximum value from list
print('Sum operation:', sum(numbers_list)) # returns a aggregate total of a list
```

## Re-ordering Lists

### Reversing Lists

```python
# using built-in method to reverse the list
numbers_list.reverse() # reverses the order of list items in place
print(numbers_list)
```

```python
# using list slicing method to reverse a list ordering
numbers_list[::-1]
```

```python
print(numbers_list) # notice how to reset the ordering of numbers_list after .reverse() call, it needs to be called again
numbers_list.reverse()
print(numbers_list)
```

```python
combined_list = my_list + numbers_list
print(combined_list)
```

## Updating List Values

### Direct positional re-assignments

```python
# changing values within my_list
print(my_list)
my_list[5] = 'Virginia' # updating the fourth position or fifth element in my_list
print(my_list)
my_list[6][2] = 'Noah' # updating the fifth position list only in the second position
print(my_list)
```
