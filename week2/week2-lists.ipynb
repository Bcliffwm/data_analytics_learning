# What is a List?
## A list is an __*ordered*__ sequence of elements that are accessible and iterable via the *index*. The index is another list that holds a positional value associated with each element of the list for retrieval and sorting purposes.

#### Lists can hold multiple primative data types and composite data types
#### List indexes are 0 indexed - meaning that you always start with 0
#### __*Lists are mutable objects*__

# Accessing Lists - Slicing and Indexing

```python
my_list = ['1', 2, 'a', 5.6, 'New York']

print(my_list) # printing the entire list
print(my_list[0]) # printing the first element of the list
print(my_list[1:3]) # printing a subset of the list
print(my_list[-1]) # printing the last element of the list
```

```python
[type(elem) for elem in my_list] # printing each element's data type using list comprehension
```

## List Slicing

* first index value is starting position in a list (inclusive)
* second index value is last position in a list (not inclusive)
* third index value is the skip value

```python
print(my_list[0:4]) # first position of list, fourth position of list
print(my_list[1:5]) # second position of list, fifth position of list
print(my_list[:3]) # grabbing everything before the 3rd element of the list (inclusive)
print(my_list[1:]) # grabbing everything after the 2nd element of the list (inclusive)
print(my_list[0:5:2]) # first position of list, fifth position of list, skip by 2
print(my_list[5:0:-1]) # fifth position of list, first position of list, skip by -1
```

```python
print(my_list.index(2)) # prints position of the value provided - in this case 1
print(my_list.index('New York'))
try:
    print(my_list.index('Virginia')) # if value provided is not in list, then ValueError is raised
except ValueError:
    print('Value is not in my_list')
```

# Useful List Methods

```python
# printing the list methods
print([prop for prop in dir(my_list) if prop[0] != '_'])
```

## Adding Values to a List

### Append Method

```python
my_list.append(['Hannibal', 'Jason', 'Ross']) # append adds element to a list at the very end - notice how lists can also be added to lists
print(my_list)
```

```python
# initializing empty list to store an array of values
numbers_list = []
curr_num = 0 # initializing int var to add values to list

# Using a while loop to append values to numbers_list
while curr_num <= 50:
    numbers_list.append(curr_num) # adding current number in the loop to list
    curr_num += 5 # adding 5 to the current number
```

### Insert Method

```python
# takes a position value and object as arguments then inserts object at specific position, updating list values and index
my_list.insert(3, 'Trade')
print(my_list)
```

### Extend Method

```python
my_list.extend(['Ben', 'is', 'amazing']) # adding elements by using an iterator
print(my_list)
```

## Mathematical Operations for lists

```python
print(numbers_list)
print('Length operation:', len(numbers_list)) # returns the count of items in a list
print('Minimum operation:', min(numbers_list)) # returns minimum value from list
print('Maximum operation:', max(numbers_list)) # returns maximum value from list
print('Sum operation:', sum(numbers_list)) # returns a aggregate total of a list
```

## Re-ordering Lists

### Sorting Lists

```python
my_fruits = ['Orange', 'Strawberry', 'Watermelon', 'Jason', 'Pineapple', 'Grapes', 'Grapefruit', 'Mango']
my_fruits.sort() # sort method orders elements by ASCII value in the case of string objs in ascending order by default
numbers_list.sort() # sort method orders int elements by value in ascending order by default
print(my_fruits)
print(numbers_list)

my_fruits.sort(reverse=True) # reverse arg orders elements in descending order
numbers_list.sort(reverse=True)
print('---- Reverse Sorting ---')
print(my_fruits)
print(numbers_list)
```

### Reversing Lists

```python
# using built-in method to reverse the list
numbers_list.reverse() # reverses the order of list items in place
print(numbers_list)
```

```python
# using list slicing method to reverse a list ordering
numbers_list[::-1]
```

```python
print(numbers_list) # notice how to reset the ordering of numbers_list after .reverse() call, it needs to be called again
numbers_list.reverse()
print(numbers_list)
```

```python
combined_list = my_list + numbers_list # overloading the addition operator to merge together two different lists
print(combined_list)
```

## Updating List Values

### Direct positional re-assignments

```python
# changing values within my_list
print(my_list)
my_list[5] = 'Virginia' # updating the fourth position or fifth element in my_list
print(my_list)
my_list[6][2] = 'Noah' # updating the fifth position list only in the second position
print(my_list)
```

## Removing List Values

#### del function

```python
# Original List
print(my_fruits)
```

```python
del my_fruits[5] # deleting an item by its index position
print(my_fruits)
```

```python
del my_fruits[my_fruits.index('Grapes')] # Deleting an object by looking up index position of item to remove
print(my_fruits)
```

#### pop method

```python
print(my_fruits)
print(my_fruits.pop()) # returns the last item in a list by default. Items returned by pop method are subsequently removed from the list
print(my_fruits)
```

```python
print(my_fruits.pop(2)) # items can also be returned by index position
print(my_fruits)
```

## List Comprehensions

#### List comprehensions are a concise way of creating lists. From Python documentation - "Common applications are to make new lists where each element is the result of some operations applied to each member of another sequence or iterable or create a subsequence of those elements that satisfy a condition"

```python
numbers_list.clear() # using clear method to keep list in memory, but remove its contents
for x in range(10):
    numbers_list.append(x * 2)
print(numbers_list)
```

#### List comprehension example

a local variable is overwritten to handle the loop processing

```python
numbers_list.clear()
numbers_list = [x * 3 for x in range(20)] # list comprehension to add values to numbers_list - in this case the var x is being overwritten
print(numbers_list)
```

```python
print('List comprehension:',[fruit.lower() for fruit in my_fruits]) # methods can be applied using list comprehensions as well
print('Combination of list and map functions:', list(map(lambda x: x.lower(), my_fruits))) # equivalent capability can be done this way, but harder to understand than list comprehension
```

## Nested Lists

A nested list is a list that exists inside of another list

```python
# In the case of my_list, the list of names ['Hannibal', 'Jason', 'Noah'] is an inner list
my_list.index(my_list)
```

```python
# using a list comprehension to find all instances of nested lists - retrieving the first result that appears
[x for x in my_list if type(x) == list][0] 
```
