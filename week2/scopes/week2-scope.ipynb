# What is Scope?

### GeekForGeeks.com - What is Scope? [website here](https://www.w3schools.com/programming/prog_scope.php)
Scope defines where variables can be accessed or modified in your program. It determines the visibility and lifetime of variables - in other words, which parts of your program can "see" and use a particular variable.

Think of scope as a container. Variables defined in one container (scope) may or may not be visible to code in other containers, depending on the type of scope.

### Python Bible pg 48
The evaluation of *global* versus *local* variables

Scopes in programming are mostly concerned with class variables, functions, conditions, and loops.

### Example from Python Bible pg 48

```python
from pprint import pprint # importing pretty print to see dict objects more cleanly
```

```python
[elem for elem in dir(globals()) if elem[0] != '_']
```

```python
# Notice how globals and locals functions generate dict objects! You can analyze your kernel or runtime's state using dict data type
print(globals().__class__)
print(locals().__class__)
```

```python
# Defining global variable
number = 15
print('All global items:')
pprint(globals().keys(), indent=2)
pprint(globals().values(), indent=2)
```

```python

print('All local items outside my_func:')
pprint(locals())

def my_func():
    my_str = 'Hello from inside my_func'
    print('All local variables inside my_func:', locals())
    print(number)

print('*' * 100)
print('Calling my_func here:')

my_func()
```

### Notice how when we call locals() from within the my_func definition, the output looks different from when we call it outside my_func
