# What is Scope?

### GeekForGeeks.com - What is Scope? [website here](https://www.w3schools.com/programming/prog_scope.php)
Scope defines where variables can be accessed or modified in your program. It determines the visibility and lifetime of variables - in other words, which parts of your program can "see" and use a particular variable.

Think of scope as a container. Variables defined in one container (scope) may or may not be visible to code in other containers, depending on the type of scope.

## Global vs Local variables

Global variables are callable and accessible to the interpreter or compiler from anywhere in the program. Local variables are only callable and accessible from within the namespace they are defined in. To make a global variable in Python, we use the *global* keyword

```python
a_number = 15

def my_new_func():
    global a_number # using global keyword to access a_number which would be out of scope without the keyword
    a_number += 10
    print('printing a_number:', a_number)

my_new_func()
my_new_func()
```

```python
b_number = 20

def new_func():
    b_number += 10
    print(b_number)

try:
    new_func()
except UnboundLocalError as e:
    print(e, '. Try updating the scope of the function, new_func', sep='')
```

Notice in the Try/Except statement that we hit the UnboundLocalError in our code. This means that we tried to access the value of a local variable before it was assigned anything.
