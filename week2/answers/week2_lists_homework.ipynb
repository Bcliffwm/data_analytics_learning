## Part C — Comprehensions and nested lists (30 points)

### Exercise 8 — List comprehensions (15 points)
Using **list comprehensions**, produce:
- `squared_evens`: the squares of only the even values in `values`
- `lowercase_fruits`: lowercase versions of all entries in `fruits`

Do not modify the original lists.

```python
values = [1, 2, 3, 4, 5, 6, 7, 8]
fruits = ["APPLE", "Banana", "GRAPE"]
squared_evens = [x**2 for x in values if x % 2 == 0]
lowercase_fruits = [fruit.lower() for fruit in fruits]
```

```python
print(squared_evens)
print(lowercase_fruits)
```

## Bonus — A mini gradebook (10 points)
Write `student_totals(gradebook)` that takes a nested list such as `[["Avery", [8, 10]], ["Jordan", [9, 10]]]` and returns a nested list with each student's name and total, e.g. `[["Avery", 18], ["Jordan", 19]]`. Use a list comprehension. The input must remain unchanged.

```python
def student_totals(gradebook):
  return [[name, sum(score)] for name, score in gradebook]
```

```python
student_totals([["Avery", [8, 10]], ["Jordan", [9, 10]]])
```
