### 1. Indexing and slicing (10 points)
Given `message = "Learning Python Strings"`, return a tuple containing (1) the first character, (2) the last character, (3) the substring `Python`, and (4) the entire message reversed. Use indexing/slicing.

```python
def inspect_message(message):
    # TODO: return (first, last, python_word, reversed_message)
    first_char, last_char = message.split(' ')[0][0], message.split(' ')[-1][-1]
    python_word = message.split(' ')[1]
    reversed_message = message[::-1]
    return (first_char, last_char, python_word, reversed_message)
```

### 4. Split and join (10 points)
Given a comma-separated string such as `"Hannibal, Ben, Jason"`, return `"Hannibal | Ben | Jason"`. Remove spaces around individual names; use `.split()`, `.strip()`, and `.join()`.

```python
def format_names(names_text):
    sep = " | "
    names = [name.strip() for name in names_text.split(',')]
    return sep.join(names)
```

```python
names = "Hannibal, Ben, Jason".split(',')
fixed_names = [name.strip() for name in names]
sep = " | "
print(sep.join(fixed_names))
```

### 7. Clean messy data (10 points)
Write `clean_person_name(raw_name)` to remove leading/trailing whitespace, lowercase all letters, and remove punctuation (including commas and exclamation marks). **Keep internal spaces.**

Example: `"   Nolan, Christopher!   "` → `"nolan christopher"`.

Hint: Your lesson used `string.punctuation` and `str.translate()`.

```python
import string

def clean_person_name(raw_name):
    # TODO
    return raw_name.strip().lower().translate(str.maketrans("", "", string.punctuation))
```

### 8. Parse a log line (10 points)
A log entry has the structure `YYYY-MM-DD HH:MM:SS LEVEL message...`.

Write `parse_log(line)` to return a dictionary with keys `timestamp`, `level`, and `message`. Example:

`"2026-10-12 16:20:05 ERROR Database connection failed"`

→ `{"timestamp": "2026-10-12 16:20:05", "level": "ERROR", "message": "Database connection failed"}`.

Hint: Use `.split(" ", 3)` to keep the message together.

```python
def parse_log(line):
    # TODO
    lines = line.split(" ", 3)
    return {"timestamp": lines[0]+' '+lines[1], "level": lines[2], "message": lines[3]}
```

## Bonus — Running time conversion (+10 points)

### 10. Convert a running time
Your lesson converted strings such as `"1:30:45"` into seconds. Write `time_to_seconds(time_text)` for a string in `H:MM:SS` format, returning the total seconds as an integer.

Example: `"1:30:45"` → `5445`. Assume the input format is valid.

```python
def time_to_seconds(time_text: str) -> int:

    seconds = 0
    # using split method on input param - iterating over each string component in list comprehension
    splits = time_text.split(':')
    converted_seconds = 0 # initializing int var for handling time conversions to seconds
    hr_min_sec = list(map(int, splits))
    reversed_time = hr_min_sec[::-1]
    for i in range(len(reversed_time)):
      seconds += (reversed_time[i] * 60 ** i)
    return seconds
```
