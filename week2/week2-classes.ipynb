# What is a Class?

## A Class is a composite data type that defines the attributes and properties of the objects that are initialized under it. In essence, the class is a *blueprint* of sorts for other objects.

Source: Python Bible pg 73

# What is an Object?

### A data object is an instance of a class read into memory. It must be initialized and physically have data stored inside

## What is an Attribute? What is a Method?

### An attribute is a variable or field that belongs to a class. A method or class function is any function defined in the class definition itself. Attributes and methods make up the definition of a class

```python
import random # importing random class for class example
```

#### Using data class example from Python Bible pg 73

```python

class Car: # starting class definition using the 'class' keyword
    # Creating a class description
    """
    A class to represent cars in inventory.

    Attributes:
    manufacturer : str
       The name of the automaker of the car
    model : str
       The name of the model of the car
    hp : int
       The horsepower the car has
    door_cnt: int
       The number of doors the car has
    convertible : bool
        Check if car is has convertible top
    price : int
        The price of the car
    foreign : bool
        Check if car is either American/Domestic or Foreign. Uses the american_makes class variable as check
    id: int
        Random integer assigned for lookup purposes of the inventory lot
    """
    amount_cars = 0 # class variable for holding number of cars
    current_ids = [] # initializing list of ids of initialized cars
    total_cost = 0 # initializing running total of all cost of cars
    # storing list in class varaibles for all American makes
    american_makes = ['Ford', 'GM', 'Cadillac', 'Dodge', 'Pontiac', 'Buick', 'Jeep', 'Chrysler', 'Ram', 'Lincoln', 'Tesla', 'Rivian', 'Lucid']
    # storing count of American makes for profiling car types
    american_make_cnt = 0

    def __init__(self, manufacturer, model, hp, door_cnt, convertible, price): # constructor method that runs each time new class instance is instanciated
        self.manufacturer = manufacturer # assigning attributes based on initialization
        self.model = model
        self.hp = hp
        self.door_cnt = door_cnt
        self.convertible = convertible
        self.price = price

        # Conditional assignments
        self.foreign = True if manufacturer not in Car.american_makes else False
        while (new_id := random.randint(1, 100)) in Car.current_ids:
            pass
        self.id = new_id

        # Class variable assignments
        Car.current_ids.append(self.id)  #adding id to tracking list
        Car.amount_cars += 1 # adding one for each new car object initialized
        Car.total_cost += self.price # adding price to total cost of inventory
        Car.american_make_cnt += 1 if self.foreign == False else 0 # adding count of domestic makes

    # Destructor method - method that gets called whenever class object gets deleted
    def __del__(self):
        print(f"Car ID {self.id} is deleted!")
        Car.amount_cars -= 1
        Car.total_cost -= self.price
        # Safely remove only if it exists (prevents crashes during partial initialization)
        if hasattr(self, "id") and self.id in Car.current_ids:
            Car.current_ids.remove(self.id)

    def print_car_info(self):
        print(
            f"Car Info\n---\nInventory ID: {self.id}\nModel: {self.model}\nManufacturer: {self.manufacturer}\nDoor Count: {self.door_cnt}\nConvertible?: {'No' if not self.convertible else 'Yes'}"
        )

    @staticmethod  # Add staticmethod decorator so it can be called without an instance
    def print_car_inventory():
        print('Car Inventory Info\n-----')
        print(f"Total Cars: {Car.amount_cars}")
        print(f"Active IDs: {Car.current_ids}")
        print(f"Total Cost of Inventory: ${Car.total_cost}.00")
        print(f"Total number of American makes: {Car.american_make_cnt}")
        print(f"Total number of Foreign makes: {Car.amount_cars - Car.american_make_cnt}")
    
```

## What is *self* and why is it used in Class definitions?

#### The self keyword is a reference on the physical memory to the class object itself. When you, the programmer, need to interact with a specific class, Python will need an indicator or reference to the exact object you need the program to interact with such as its initialization or deletion.

### Constructors and Destructors
A constructor is a function defined inside of a class that executes whenever a new instantiation of a class occurs. Constructors accept n number of parameters depending on the attributes of the class definition. However, the most important parameter of any constructor is the __self__ parameter - which makes a reference back to the object. 

In the Car object, the constructor accepts several parameters to initialize into attributes for objects read into memory such as price, car door count, and other parameters. Each Car object has access to the Car class variables such as amount_cars, current_ids, total_cost, american_makes, and american_make_cnt.

A destructor is a function defined inside of a class that executes whenever a class object gets deleted or removed from runtime memory. Destructors only need the __self__ parameter to be defined (but could accept other params).

In the Car object, the destructor prints message to the terminal and then accesses the class variables controlling an inventory list of the Car objects.

```python
a_car = Car("Tesla", "Model X", 525, 4, False, 150000) # initializing a Car class
a_car.print_car_info()
print()
print()
Car.print_car_inventory() # using the staticmethod defined in the Car class
```
