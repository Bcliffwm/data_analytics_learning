# What is a database?

### A database is a structured system for efficient storage, retrieval, organization, and modification enabling simple to complex applications or systems with predictable and scalable data hydration

[**(Source: Microsoft Azure - What are databases?)**](https://azure.microsoft.com/en-us/resources/cloud-computing-dictionary/what-are-databases)

```python
import sqlite3
```

```python
conn = sqlite3.connect('students.db')
```

```python
try:
    cursor = conn.cursor()

    cursor.execute(""" 
    CREATE TABLE IF NOT EXISTS students (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        gpa FLOAT NOT NULL,
        major TEXT
    );
    """)

    # Adding a new student record
    new_student = ("Ben Cliff", 3.3, "Information Technology")

    # Executing SQL command
    cursor.execute(
        "INSERT INTO students (name, gpa, major) VALUES (?, ?, ?)",
        new_student)

    # Adding multiple in list object
    new_students = [("Hannagirl", 3.5, "Data Science"),
                    ("JSON", 3.2, "Ross's hobbies idk")]
    
    # Executing a multi-row insert into DB
    cursor.executemany("INSERT INTO students (name, gpa, major) VALUES (?, ?, ?)",
                       new_students) 
    
    conn.commit() # Saving all writes to the database file
except sqlite3.Error as e:
    print(f"An error occurred when inserting information: {e}")
# Always shutting down the connection to db once transaction is complete
finally:
    conn.close()
```

## Information Retrieval Options

```python
# Attempting to retrieve records from database
conn = sqlite3.connect('students.db') # need to re-establish connection to database
try:
    cursor = conn.cursor() # initializing another cursor object to interface with database

    # Retrieving information from the database
    print("Database SELECT * results")
    print('*' * 30)
    results = cursor.execute("SELECT * FROM students") # records are returned as a Cursor object
    
    for row in results: # iterating through each of the results
        print(row)
    print("The data type for results:", type(results))
    [print(method) for method in dir(results) if method[0] != '_']
    print("\n")
    filters = cursor.execute("SELECT name, gpa FROM students WHERE gpa >= 3.3")

    print('Database Filtered results')
    print('*' * 30)
    
    for row in filters:
        print(row)
except sqlite3.Error as e:
    print(f"An error occurred when retrieving information: {e}")
finally:
    conn.close()
```

```python
# Attempting to retrieve records from database
conn = sqlite3.connect('students.db') # need to re-establish connection to database
try:
    cursor = conn.cursor() # initializing another cursor object to interface with database

    # Retrieving information from the database
    print("Database SELECT * results")
    print('*' * 30)
    results = cursor.execute("SELECT * FROM students") # records are returned as a Cursor object
    
    print(results.fetchall())
except sqlite3.Error as e:
    print(f"An error occurred when retrieving information: {e}")
finally:
    conn.close()
```
